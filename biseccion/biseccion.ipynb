```python
from numba.core.types import none

#Metodo de la biseccion
def biseccion(f, a, b, tolerancia =1e-10, max_iter=1000, print_table=False ):
    if f(a) * f(b) >=0:
        print("El metodo de biseccion no puede encontrar una raiz en el inventario dado. ")
        return None

    for _ in range(max_iter):
        c = (a+b)/2
        if (f(c) ==0) or abs(b-a) < tolerancia:
            return c, _
        if f(a) * f(c)<0:
            b=c
        elif f(b) * f(c) < 0:
            a=c
        if print_table:
            print(_, "|", a, "|", b, "|", c,"|", f(c))
    return c, max_iter
```
